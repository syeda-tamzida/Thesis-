```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# Load the grayscale image
image = cv2.imread('/content/100.jpg', cv2.IMREAD_GRAYSCALE)

# Check if the image was loaded
if image is None:
    print("Error: Image not loaded. Check the file path.")
else:
    # Apply Gaussian blur to reduce noise
    blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

    # Sobel edge detection
    sobel_x = cv2.Sobel(blurred_image, cv2.CV_64F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(blurred_image, cv2.CV_64F, 0, 1, ksize=3)

    # Gradient magnitude (edge strength)
    gradient_magnitude = cv2.magnitude(sobel_x, sobel_y)
    gradient_magnitude = cv2.convertScaleAbs(gradient_magnitude)

    # Threshold to keep strong edges
    _, edge_map = cv2.threshold(gradient_magnitude, 50, 255, cv2.THRESH_BINARY)

    # Canny edge detection for comparison
    canny_edges = cv2.Canny(blurred_image, 100, 200)

    # Surface roughness (standard deviation of gradient magnitudes)
    roughness = np.std(gradient_magnitude)

    # Display results
    print(f"Surface Roughness (Standard Deviation): {roughness:.2f}")

    print("\nGradient Magnitude (Sobel):")
    cv2_imshow(gradient_magnitude)

    print("\nThresholded Edge Map:")
    cv2_imshow(edge_map)

    print("\nCanny Edge Detection:")
    cv2_imshow(canny_edges)

    # Plot histogram of edge magnitudes
    plt.figure(figsize=(6, 4))
    plt.hist(gradient_magnitude.ravel(), bins=50, color='gray')
    plt.title("Histogram of Gradient Magnitudes")
    plt.xlabel("Intensity")
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')
X = np.arange(0, gradient_magnitude.shape[1])
Y = np.arange(0, gradient_magnitude.shape[0])
X, Y = np.meshgrid(X, Y)
Z = gradient_magnitude

ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')
ax.set_title('3D Surface Plot of Gradient Magnitude')
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Edge Intensity')
plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import pandas as pd
from google.colab import files
import os

# Step 1: Upload images
uploaded = files.upload()

# Step 2: Initialize results list
results = []

# Step 3: Loop through uploaded files
for filename in uploaded.keys():
    # Load image in grayscale
    image = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)

    if image is None:
        print(f"Error: Could not load {filename}")
        continue

    # Apply Gaussian blur
    blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

    # Sobel edge detection
    sobel_x = cv2.Sobel(blurred_image, cv2.CV_64F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(blurred_image, cv2.CV_64F, 0, 1, ksize=3)
    gradient_magnitude = cv2.magnitude(sobel_x, sobel_y)
    gradient_magnitude = cv2.convertScaleAbs(gradient_magnitude)

    # Surface roughness as standard deviation
    roughness = np.std(gradient_magnitude)

    # Add to results
    results.append({'Image Name': filename, 'Surface Roughness': round(roughness, 2)})

# Step 4: Create DataFrame and save to Excel
df = pd.DataFrame(results)
excel_filename = 'surface_roughness_results.xlsx'
df.to_excel(excel_filename, index=False)

# Step 5: Download Excel file
files.download(excel_filename)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Load grayscale image
image_path = '/content/100.jpg'
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

if image is None:
    print("❌ Error: Image not loaded. Check the file path.")
else:
    # Preprocessing
    blurred = cv2.GaussianBlur(image, (5, 5), 0)
    sobel_x = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)
    gradient = cv2.magnitude(sobel_x, sobel_y)
    gradient = cv2.convertScaleAbs(gradient)

    # 1️⃣ Histogram
    plt.figure(figsize=(6, 4))
    plt.hist(gradient.ravel(), bins=50, color='gray')
    plt.title("Histogram of Gradient Magnitudes")
    plt.xlabel("Intensity")
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("/content/histogram_gradient_magnitude.png", dpi=1000)
    plt.close()

    # 2️⃣ 3D Surface Plot
    fig = plt.figure(figsize=(10, 6))
    ax = fig.add_subplot(111, projection='3d')
    X = np.arange(0, gradient.shape[1])
    Y = np.arange(0, gradient.shape[0])
    X, Y = np.meshgrid(X, Y)
    ax.plot_surface(X, Y, gradient, cmap='viridis', edgecolor='none')
    ax.set_title('3D Surface Plot of Gradient Magnitude')
    ax.set_xlabel('X')
    ax.set_ylabel('Y')
    ax.set_zlabel('Intensity')
    plt.tight_layout()
    plt.savefig("/content/3d_surface_gradient_magnitude.png", dpi=1000)
    plt.close()

    # 3️⃣ Line Profile
    row = gradient.shape[0] // 2
    plt.figure(figsize=(8, 4))
    plt.plot(gradient[row], color='blue')
    plt.title(f"Intensity Profile Along Row {row}")
    plt.xlabel("Pixel Index")
    plt.ylabel("Gradient Intensity")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("/content/line_profile_gradient.png", dpi=1000)
    plt.close()

    # 4️⃣ Boxplot
    plt.figure(figsize=(4, 6))
    plt.boxplot(gradient.ravel(), vert=True)
    plt.title("Boxplot of Gradient Intensities")
    plt.ylabel("Intensity")
    plt.grid(True)
    plt.tight_layout()
    plt.savefig("/content/boxplot_gradient_intensities.png", dpi=1000)
    plt.close()

    print("✅ All plots saved at 1000 DPI in /content/")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

# Load the grayscale image
image = cv2.imread('/content/7.jpg', cv2.IMREAD_GRAYSCALE)

# Check if the image was loaded
if image is None:
    print("Error: Image not loaded. Check the file path.")
else:
    # Display original image
    print("Original Image:")
    cv2_imshow(image)

    # Apply Gaussian blur to reduce noise
    blurred_image = cv2.GaussianBlur(image, (5, 5), 0)
    print("\nBlurred Image (Gaussian 5x5):")
    cv2_imshow(blurred_image)

    # Sobel edge detection
    sobel_x = cv2.Sobel(blurred_image, cv2.CV_64F, 1, 0, ksize=3)
    sobel_y = cv2.Sobel(blurred_image, cv2.CV_64F, 0, 1, ksize=3)

    # Convert Sobel outputs to absolute values and 8-bit
    sobel_x_abs = cv2.convertScaleAbs(sobel_x)
    sobel_y_abs = cv2.convertScaleAbs(sobel_y)

    # Display Sobel components
    print("\nSobel X (Horizontal Edges):")
    cv2_imshow(sobel_x_abs)

    print("\nSobel Y (Vertical Edges):")
    cv2_imshow(sobel_y_abs)

    # Create color-coded Sobel edges
    sobel_combined = cv2.addWeighted(sobel_x_abs, 0.5, sobel_y_abs, 0.5, 0)
    sobel_color = cv2.applyColorMap(sobel_combined, cv2.COLORMAP_JET)
    print("\nColor-coded Sobel Edges (Heatmap):")
    cv2_imshow(sobel_color)

    # Gradient magnitude (edge strength)
    gradient_magnitude = cv2.magnitude(sobel_x, sobel_y)
    gradient_magnitude = cv2.convertScaleAbs(gradient_magnitude)
    print("\nGradient Magnitude (Sobel):")
    cv2_imshow(gradient_magnitude)

    # Create heatmap of gradient magnitude
    gradient_heatmap = cv2.applyColorMap(gradient_magnitude, cv2.COLORMAP_JET)
    print("\nGradient Magnitude Heatmap:")
    cv2_imshow(gradient_heatmap)

    # Threshold to keep strong edges
    _, edge_map = cv2.threshold(gradient_magnitude, 50, 255, cv2.THRESH_BINARY)
    print("\nThresholded Edge Map (Binary):")
    cv2_imshow(edge_map)

    # Color-coded thresholded edges
    edge_heatmap = cv2.applyColorMap(edge_map, cv2.COLORMAP_JET)
    print("\nThresholded Edge Heatmap:")
    cv2_imshow(edge_heatmap)

    # Canny edge detection for comparison
    canny_edges = cv2.Canny(blurred_image, 100, 200)
    print("\nCanny Edge Detection:")
    cv2_imshow(canny_edges)

    # Color-coded Canny edges
    canny_heatmap = cv2.applyColorMap(canny_edges, cv2.COLORMAP_JET)
    print("\nCanny Edge Heatmap:")
    cv2_imshow(canny_heatmap)

    # Surface roughness (standard deviation of gradient magnitudes)
    roughness = np.std(gradient_magnitude)
    print(f"\nSurface Roughness (Standard Deviation): {roughness:.2f}")

    # Plot histogram of edge magnitudes
    plt.figure(figsize=(10, 6))
    plt.hist(gradient_magnitude.ravel(), bins=50, color='gray')
    plt.title("Histogram of Gradient Magnitudes")
    plt.xlabel("Intensity")
    plt.ylabel("Frequency")
    plt.grid(True)
    plt.show()

    # Additional visualizations
    plt.figure(figsize=(15, 10))

    # Original and blurred
    plt.subplot(2, 3, 1)
    plt.imshow(image, cmap='gray')
    plt.title('Original Image')

    plt.subplot(2, 3, 2)
    plt.imshow(blurred_image, cmap='gray')
    plt.title('Blurred Image')

    # Sobel components
    plt.subplot(2, 3, 3)
    plt.imshow(sobel_x_abs, cmap='gray')
    plt.title('Sobel X')

    plt.subplot(2, 3, 4)
    plt.imshow(sobel_y_abs, cmap='gray')
    plt.title('Sobel Y')

    # Combined results
    plt.subplot(2, 3, 5)
    plt.imshow(cv2.cvtColor(sobel_color, cv2.COLOR_BGR2RGB))
    plt.title('Color Sobel')

    plt.subplot(2, 3, 6)
    plt.imshow(gradient_magnitude, cmap='jet')
    plt.title('Gradient Magnitude')
    plt.colorbar()

    plt.tight_layout()
    plt.show()
```
