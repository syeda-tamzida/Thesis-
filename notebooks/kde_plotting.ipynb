```python
import pandas as pd
import seaborn as sns

df = pd.read_csv('data.csv')

df.dropna(inplace=True)
df
```

**temperature**

```python
import matplotlib.font_manager as fm
import matplotlib as mpl

font_path = r"C:\Users\LENOVO\AppData\Local\Microsoft\Windows\Fonts\HELVETICA.TTF"
fm.fontManager.addfont(font_path)
prop = fm.FontProperties(fname=font_path)
font_name = prop.get_name()

font_name
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import matplotlib as mpl

# Set Helvetica font globally
mpl.rcParams['font.family'] = 'Helvetica'
mpl.rcParams['pdf.fonttype'] = 42      # Embeds fonts properly for journals
mpl.rcParams['ps.fonttype'] = 42

fig, ax = plt.subplots(figsize=(7, 5))  # ~single column journal width

sns.kdeplot(
    data=df,
    x='temperature_c',
    fill=True,
    color='#2d6a2d',
    facecolor='#8fbc8f',
    alpha=0.6,
    linewidth=1.5,
    ax=ax
)

# Axis labels — journal style
ax.set_xlabel('Temperature (°C)', fontsize=10, labelpad=6)
ax.set_ylabel('Density', fontsize=10, labelpad=6)

# Tick formatting
ax.tick_params(axis='both', labelsize=9, direction='out', length=4, width=0.8)

# Clean spines — keep only bottom and left
# ax.spines['top'].set_visible(False)
# ax.spines['right'].set_visible(False)
ax.spines['left'].set_linewidth(0.8)
ax.spines['bottom'].set_linewidth(0.8)

# White background
fig.patch.set_facecolor('white')
ax.set_facecolor('white')

plt.tight_layout()

# Save as high-res for journal submission
plt.savefig('temp_kde_plot.pdf', dpi=1000, bbox_inches='tight', format='pdf')  # PDF preferred
plt.show()
```
